```python
#Question 1.3 --- Identify Missing Values 

for name, df in all_datasets.items():

    print(f" Missing Values in {name}")
    missing_count = df.isnull().sum()
    print(missing_count[missing_count >0] if missing_count.sum()>0 else "No missing values found")
    print()
```

```python
#Calculate the percentage of missing value in each column

for name, df in all_datasets.items():

    print(f" % Missing values in {name}")

    missing_pct = (df.isnull().sum())/len(df)*100

    missing_pct = missing_pct[missing_pct>0] ## keep only columns that actually have missing values.

    if len(missing_pct) == 0:
        print("No missing values found")

    else:
        print((missing_pct).round(2))

    print()
```

```python
#highlight which column has the highest percentage of missing values in each dataset.

for name, df in all_datasets.items():

    missing_pct = df.isnull().sum() / len(df) * 100

    if missing_pct.max() == 0:
        print(f"{name}: No missing values")
    else:
        worst_column = missing_pct.idxmax()
        worst_value = missing_pct.max()

        print(
            f"{name}: highest missing percentage is "
            f"'{worst_column}' with {worst_value:.2f}%"
        )
```

```python
#----------Question 3-- Data Qualtiy Checks---------

#Question 3.1 ----- For each state calculate the total number of transactions, 
#total transaction amount and total registered users by summing up the values from the district level data.


district_summary = district_txn_users.groupby(["State", "Year","Quarter"])[[
    "Transactions",
    "Amount (INR)",
    "Registered Users"
]].sum().reset_index()

display(district_summary)

                                     
```

```python
comparison = district_summary.merge(state_txn_users, on=["State", "Year", "Quarter"],
    suffixes=("_District", "_State")
)

display(comparison)
```

```python
comparison["Transaction_Difference"] = (
    comparison["Transactions_District"] -
    comparison["Transactions_State"]
)

comparison["Amount_Difference"] = (
    comparison["Amount (INR)_District"] -
    comparison["Amount (INR)_State"]
)

comparison["Users_Difference"] = (
    comparison["Registered Users_District"] -
    comparison["Registered Users_State"]
)


discrepancies = comparison[
    (comparison["Transaction_Difference"] != 0) |
    (comparison["Amount_Difference"] != 0) |
    (comparison["Users_Difference"] != 0)
]

display(discrepancies)
```

```python
#Question 4.2 -------------Correlate population denstiy with transaction volume. 

#Question 4.2.1---------

#Merge the District Txn and user dataset with the District Demographics.


density_data = district_demographics[
    ["State", "District", "Density"]
].copy()

transaction_data = (
    district_txn_users
    .groupby(["State", "District"])["Transactions"]
    .sum()
    .reset_index()
)

Merge_Data_DTU_DD = pd.merge(
    density_data,
    transaction_data,
    on=["State", "District"],
    how="inner"
)

display(Merge_Data_DTU_DD)
```

```python
#Question 4.2.2-------Calculate the Correlation between population density and transaction volume. 

Correlation = (
    Merge_Data_DTU_DD["Density"]
    .corr(Merge_Data_DTU_DD["Transactions"])
)

print("Correlation:", round(Correlation, 2))
```

```python
#Question 4.4.2-----------Calculate the ratio of users using each device brand,
                           # to the total number of registered users in each state.
                           #Display the result in tabular form. 


Device_brand_state = (
    state_devicedata
    .groupby(["State", "Brand"])["Registered Users"]
    .sum()
    .reset_index()
)

state_total_users = (
    Device_brand_state
    .groupby("State")["Registered Users"]
    .transform("sum")
)

Device_brand_state["State Brand Ratio User"] = (
    Device_brand_state["Registered Users"] /
    state_total_users
)

display(Device_brand_state)
```

```python
#Question 6.2--- Find Correlations between demographic data (e.g. population density) and transaction data (e.g. transaction volume). 
#Summarise your finding.

# 1. Get density for each district
density_data = district_demographics[
    ["State", "District", "Density"]
].copy()

# 2. Calculate total transactions for each district
transaction_data = district_txn_users.groupby(
    ["State", "District"]
)["Transactions"].sum().reset_index()

# 3. Merge demographic and transaction data
correlation_data = pd.merge(
    density_data,
    transaction_data,
    on=["State", "District"],
    how="inner"
)

# 4. Remove missing values
correlation_data = correlation_data.dropna(
    subset=["Density", "Transactions"]
)

# 5. Calculate correlation
correlation = correlation_data["Density"].corr(
    correlation_data["Transactions"]
)

print("Correlation between Population Density and Transaction Volume:",
      round(correlation, 2))

# 6. Display the data
display(correlation_data)
```

```python
#Question 6.2 -- Summarise your Finding

#Answer - There is a moderate positive correlation (0.42) between population density and transaction volume at the district level. 
         #This suggests that districts with higher population density generally tend to have higher transaction volumes. 
         #However, the relationship is not very strong, indicating that other factors besides population density also influence transaction activity.
```
